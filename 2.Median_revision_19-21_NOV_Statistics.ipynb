# Median

```python
# print position of list
L = [1,4,7,9,11]

print(L[0])
print(L[1])
print(L[2])
print(L[3])
print(L[4])
```

```python
#mid position(index) of length(odd number)
L = [1,4,7,9,11]
N = len(L)
mid_pos = N//2
print("index no",mid_pos)
```

```python
# mid value/median
L = [1,4,7,9,11]
N = len(L)
mid_pos = N//2
print("median is",L[mid_pos])
```

```python
# print middle value / median (Odd number)
k = [2,4,5,6,8,9,12,14,15,17,19]
N = len(k)
mid_pos=N//2
print("median is",k[mid_pos])
```

```python
# print middle value / median 
m = [2,4,6,8]
print(m[0])
n = len(m)
print(n)
mid_pos = n//2,n//2-1
print(mid_pos)
```

```python
m = [2,4,6,8]
n = len(m)
median = (m[(n//2 - 1)])
med = (m[n//2])
print(med)
print(median)
f_median=(med+median)/2
print(f_median)
```

```python
# print middle value / median (Odd number)
k = [12,14,15,16,18,19,22,24,25,27,29,39,40,42,45,47,49]
n = len(k)
print(n)
print(k[0:18])
midpos = n//2
print("mid_pos",midpos)
print("median is",k[midpos])
```

```python
# print middle value / median (Odd number)
m = [12,14,15]
print(m[0:3])
p = len(m)
print(p)
midpos = p//2
print("mid_pos",midpos)
print("median is",m[midpos])
```

```python
# print middle value / median (Odd number)
N = [12,14,15,16,17]
print(N[0:])
n = len(N)
midpos = n//2
print("mid_pos",midpos)
print("median is",N[midpos])
```

```python
# print middle value / median (even number)
p = [12,14,15,16]
print(p[0:])
n = len(p)
midpos1,midpos2 = n//2 , n//2-1
print(midpos1,midpos2)
print("middle element is",p[midpos1],p[midpos2])
```

```python
# print middle value / median (even number)
q = [12,14,15,16,18,19]
print(q[0],q[1],q[2],q[3],q[4],q[5])
print(q[0:])
n = len(q)
midpos1,midpos2=n//2,n//2-1
print("middle index is",midpos1,midpos2)
print("middle element is",q[midpos1],q[midpos2])
```

```python
# print middle value / median (even number)
r = [12,14,15,16,18,19,20,22]
print(r[0:])
n = len(r)
midpos1 = n//2-1
midpos2 = n//2
print("midpos is",midpos1,midpos2)
print("middle element is",r[midpos1],r[midpos2])
```
